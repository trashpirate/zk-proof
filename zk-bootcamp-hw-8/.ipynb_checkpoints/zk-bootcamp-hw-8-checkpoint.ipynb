```python
# Problem 1
import galois
import numpy as np
import random
p = 103
GF = galois.GF(p)

xs = GF(np.array([1,2,3]))

# arbitrary vectors
v1 =  GF(np.array([4,8,19]))
v2 =  GF(np.array([4,4,19]))


def L(v):
    return galois.lagrange_poly(xs, v)

p1 = L(v1)
p2 = L(v2)


u = random.randint(0, p)

lhs = p1(u)
rhs = p2(u)

# only one check required
assert lhs != rhs
```

```python
# Problem 2
import galois
import numpy as np
import random

p = 103
GF = galois.GF(p)

def verify(A,B,v):
    
    (n, m) = A.shape
    print(f"rows = %d, cols = %d" % (n, m))

    xs = GF(np.arange(1,n+1))
    a = GF(0)
    b = GF(0)
    for i in range(m):
        a += v[i] * galois.lagrange_poly(xs, A[:,i])
        # print("a:", A[:,i])
        b += v[i] * galois.lagrange_poly(xs, B[:,i])
        # print("b:",B[:,i])
        
    u = random.randint(0, p)
    return a(u) == b(u)
        

# arbitrary matrices
Alice =  GF(np.matrix([[9,4,5],[8,3,4],[7,9,11]]))
Bob =  GF(np.matrix([[2,4,6],[1,3,5],[7,9,11]]))
v = GF(np.array([1,3,7])[:,np.newaxis])

verify(Alice,Bob,v)
```
