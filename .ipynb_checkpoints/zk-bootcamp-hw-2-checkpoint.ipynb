```python
# Problem 1
# Let our set be real numbers. Show a binary operator that is not closed

# Solution 1
# The square root is a binary operator that is not closed because:

import numpy as np
result = np.sqrt(complex(-1))
print(f"Square root of -1 is imaginary (not a real number): {result}")
```

```python
# Problem 2
# Find a binary operator that is closed but not associative for real numbers

# Solution 2
# Substraction is closed but not associative

import numpy as np
a = 3
b = 5
c = 1

print(f"(a - b) - c = {(a-b)-c}")
print(f" a - (b - c) = {a-(b-c)}")
```

```python
# Problem 3

# What algebraic structure (group, monoid, semigroup, etc) is all even integers under addition.

# Solution 3
# - Magma: set is closed
# - Semigroup: binary operator is associative
# - Monoid: set has an identity element
# - Group: if each element has an inverse
#     - Finite group: set is finite
#     - Abelian group: commutative
    
# Integers under addition:|
# - closed: even + even = even
# - associative: (2 + 4) + 6 = 2 + (4 + 6)
# - identity: a + 0 = a => identity element is 0
# - inverse exists: a => -a
print("Integers under addition are a abelian group.")
```

```python
# Problem 4

# What algebraic structure is all odd integers under multiplication?

# Solution 4
# - Magma: set is closed
# - Semigroup: binary operator is associative
# - Monoid: set has an identity element
# - Group: if each element has an inverse
#     - Finite group: set is finite
#     - Abelian group: commutative
    
# Integers under multiplication:
# - closed: odd * odd = odd
# - associative: (3 * 5) * 1 = 3 * (5 * 1)
# - identity: a * 1 = a => identity element is 1
# - inverse exists: no, inverse of a = 1/a, while 1/a is not part of the set
print("Integers under multiplication are a commutative Monoid.")
```

```python
# Problem 5

# Let our set be 3 x 2 matrices of integers under addition. What algebraic structure is this?

# Solution 5   
# Integers under multiplication:
# - closed: A + B = C where C is also a matrix of integers
# - associative: (A + B) + C = A + (B + C)
# - identity: A + 0 = A
# - inverse exists: A = -A
print("3 x 2 matrices of integers under addition are a abelian group.")
```

```python
# Problem 6

# Suppose our set is all rational numbers Q except 0 and our binary operator is division. What algebraic structure is this?

# Solution 6   
# Q under division:
# - closed: a / b = c where c is element of Q
# - associative: (a / b) / c != a / (b / c)
# - identity: N/A
# - inverse exists: N/A
print("Q under division is a Magma.")
```
