```python
# Homework 1
p = 71
```

```python
# Problem 1
'''
Find the elements in a finite field that are congruent to the following values:
- -1
- -4
- -160
- 500
'''
print(-1 % 71)
print(-4 % 71)
print(-160 % 71)
print(500 % 71)
```

```python
# Problem 2
''' 
Find the elements that are congruent to a = 5/6, b = 11/12, and c = 21/12
Verify your answer by checking that a + b = c (in the finite field)
'''
a = 5 * pow(6, -1, p) % p
b = 11 * pow(12, -1, p) % p
c = 21 * pow(12, -1, p) % p
print("a =", a)
print("b =", b)
print("c =", c)
print(f"Proof: {a} + {b} = {(a + b) % 71}") 
```

```python
# Problem 3
'''
Find the elements that are congruent to a = 2/3, b = 1/2, and c = 1/3.
Verify your answer by checking that a * b = c (in the finite field)
'''
a = 2 * pow(3, -1, p) % p
b = pow(2, -1, p) % p
c = pow(3, -1, p) % p
print("a =", a)
print("b =", b)
print("c =", c)
print(f"Proof: {a} x {b} = {(a * b) % 71}") 
```

```python
# Problem 4
'''
Compute the inverse of the following matrix in the finite field:
A = [ [1, 1], [1, 4]]
Verify your answer by checking that:
A * A**(-1) = I (identity)
'''
import numpy as np

a = 1
b = 1
c = 1
d = 4
A = np.array([[a,b],[c,d]])
Agf = A % p

det = np.linalg.det(A) # = 3.0000000000000004
det = a * d - b * c

AIgf = pow(det,-1,p) * np.array([[d % p, -b % p],[-c % p, a % p]]) % p


print("a =", a)
print("b =", b)
print("c =", c)
print("d =", d)

print("A =\n", A)
print("Det:", det)

print("\nFinite Field:")
print("Agf =\n", Agf)

print("AIgf =\n", AIgf)

# AAgf = A[0][0] * Agf[0][0] + A[0

print(f"Proof: Agf.AIgf \n= {Agf.dot(AIgf) % p}") 
```

```python
# Problem 5
'''
What is the modular square root of 12?
Verify your answer by checking that x * x = 12 (mod 71)
Use brute force to find the answer (in Python)
'''

for z in range(0,2*p):
    sq = z * z
    remainder = sq % p
    if (remainder == 12):
        print("Found: x =", z)
        print("y = x * x =", sq)
        print("y % p =", sq % p)
        print()
```

```python
# Problem 6
'''
Feel free to use an AI to help you answer this question. The goal is just to get you familiar with the [galois](https://mhostetter.github.io/galois/latest) Python library.
Suppose we have the following polynomials:

p(x)=52x^2+24x+61
q(x)=40x^2+40x+58

What is p(x) + q(x)? What is p(x) * q(x)?

p(x) + q(x) = 90x^2 + 64x + 119
p(x) * q(x) = (52x^2+24x+61)*(40x^2+40x+58) = 2080x^4+3040x^3+6416x^2+3832x+3538

Use the `galois` library in Python to find the roots of p(x) and q(x).

Here is the documentation: https://mhostetter.github.io/galois/latest/api/galois.Poly.Roots/#returns

What are the roots of p(x)q(x)?
'''
import galois
import numpy as np

GF = galois.GF(p)

# Define the polynomials p(x) and q(x) over this field
# The coefficients are automatically reduced modulo 71
px = galois.Poly([52, 24, 61], field=GF)
qx = galois.Poly([40, 40, 58], field=GF)
pqx = px * qx

# Find the roots of each polynomial in GF(71)
px_roots = px.roots()
qx_roots = qx.roots()
pqx_roots = pqx.roots()

print(f"p(x) = {px}")
print(f"Roots of p(x) in GF(71): {px_roots}\n")

print(f"q(x) = {qx}")
print(f"Roots of q(x) in GF(71): {qx_roots}\n")

print(f"p(x)q(x) = {pqx}")
print(f"Roots of p(x)q(x) in GF(71): {pqx_roots}")

```

```python
# Problem 7
'''
Find a polynomial f(x) that crosses the points (10, 15), (23, 29).
Since these are two points, the polynomial will be of degree 1 and be the equation for a line (y = ax + b).
Verify your answer by checking that f(10) = 15 and f(23) = 29.
'''
import numpy as np

a = np.array([10, 15])
b = np.array([23, 29])

delta = b - a
m = delta[1] / delta[0]

x = a[0]
y = a[1]

# y = x * m + c
c = y - x * m

def f(x):
    return x * m + c

print("f(10) =", f(10))
print("f(23) =", f(23))
```
