```python
# 1) Pick a private key
import os
from py_ecc.bn128 import G1, G2, add, multiply, curve_order, pairing, neg, field_modulus

print("Curve order:", curve_order)
print("G1:", G1)
print("G2:", G2)
```

```python
a = 6
b = 2
c = 3
d = 4
# aG1 * bG2 == cG1 * dG2

A = multiply(G1, a)
print("A:", A)
B = multiply(G2, b)
print("B:", B)
C = multiply(G1, c)
print("C:", neg(C))
D = multiply(G2, d)
print("D:", D)

pairing(B, A) * pairing(D, neg(C))
```

```python
# arguments: A_1, B_2, C_1, x_1,x_2,x_3
x1 = 2
x2 = 3
x3 = 4

A1 =  multiply(G1, 4)
print("A1:", A1)
B2 =  multiply(G2, 9)
print("B2:", B2)
C1 =  multiply(G1, 3)
print("C1:", C1)

X1 = add(add(multiply(G1, x1),  multiply(G1, x2)),  multiply(G1, x3))

# 0 = -A1B2 + Alpha1*Beta2 + X1*Gamma2 + C1*Delta2
#      -a*b + alpha*beta + x*gamma + c*delta
#        -(4) * (9) +  2 * 6 + 9 * 1 + (3) * 5 
#        -(4) * (9) + 12 + 9 + 15

Alpha1 =  multiply(G1, 2)
print("Alpha1:", Alpha1)
Beta2 =  multiply(G2, 6)
print("Beta2:", Beta2)
Gamma2 =  multiply(G2, 1)
print("Gamma2:", Gamma2)
Delta2 =  multiply(G2, 5)
print("Delta2:", Delta2)

pairing(B2, neg(A1)) * pairing(Beta2, Alpha1) * pairing(Gamma2, X1) * pairing(Delta2, C1)
```
